### Task 7

**New trick.** Format numbers inside an f string. Print these two lines:

**Hint:** `:.2f` keeps 2 decimal places. `:,` adds commas to big numbers.

### Task 9

Your hero buys tomatoes and plantain. Work out the total and print it as `Total: 4500 XAF`.

Tomatoes cost 350 each and she buys 6. Plantain costs 1200 each and she buys 2.

**Hint:** Multiply first, then add. Python follows the order of operations.

````python
tomatoes_price = 350
tomatoes_qty = 6
plantain_price = 1200
plantain_qty = 2

# Write your code below this line
total_tomatoes = tomatoes_price * tomatoes_qty
total_plantain = plantain_price * plantain_qty
total_price = total_tomatoes + total_plantain

print(f"Total: {total_price} XAF")


# Expected output:
# Total: 4500 XAF
````

## Level 5: The Boss Challenge

One big receipt. Use variables, f strings, arithmetic and concatenation together.

### Task 20 · 8 points

**Boss challenge.** Print this receipt. Use `'=' * 20` and `'-' * 20` for the lines, f strings for the item lines, and arithmetic for all the numbers. Use `.center(20)` for the title.

```text
====================
   MARKET RECEIPT   
====================
Shopper: Chantal
Tomatoes: 6 x 350 = 2100 XAF
Plantain: 2 x 1200 = 2400 XAF
--------------------
Subtotal: 4500 XAF
Discount (10%): 450.0 XAF
Total: 4050.0 XAF
Paid: 5000 XAF
Change: 950.0 XAF
====================
```

**Hint:** Work out the numbers first (line totals, subtotal, discount, total, change). Then do the printing.

````python
shopper = 'Chantal'
item1 = 'Tomatoes'
price1 = 350
qty1 = 6
item2 = 'Plantain'
price2 = 1200
qty2 = 2
discount_rate = 0.10
paid = 5000

# Write your code below this line
title = "MARKET RECEIPT"
total_tomatoes = qty1 * price1
total_plantain = qty2 * price2
total_items = total_tomatoes + total_plantain
discount = total_items * discount_rate
total_price = total_items - discount

print(f"""
{"=" * 20}
{title.center(20)}
{"=" * 20}
Shopper: {shopper}
{item1}: {qty1} x {price1} = {total_tomatoes} XAF
{item2}: {qty2} x {price2} = {total_plantain} XAF
{"-" * 20}
Subtotal: {total_items} XAF
Discount (10%): {discount} XAF
Total: {total_price} XAF
Paid: {paid} XAF
Change: {paid - total_price} XAF
{"=" * 20}
""")


# Expected output:
# ====================
#    MARKET RECEIPT
# ====================
# Shopper: Chantal
# Tomatoes: 6 x 350 = 2100 XAF
# Plantain: 2 x 1200 = 2400 XAF
# --------------------
# Subtotal: 4500 XAF
# Discount (10%): 450.0 XAF
# Total: 4050.0 XAF
# Paid: 5000 XAF
# Change: 950.0 XAF
# ====================
````
